```python
import math
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy
```

```python
class SelfAttention(nn.Module):
    def __init__(self, embed_dim):
        super().__init__()
        self.embed_dim = embed_dim

        # Define the query, key, and value linear layers
        self.query = nn.Linear(embed_dim, embed_dim, bias=False)
        self.key = nn.Linear(embed_dim, embed_dim, bias=False)
        self.value = nn.Linear(embed_dim, embed_dim, bias=False)


    def forward(self, x):
        # Compute the query, key, and value matrices
        Q = self.query(x)  
        K = self.key(x)    
        V = self.value(x)  

        # Compute the attention scores
        attn_scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.embed_dim)  # (batch_size, seq_len, seq_len)

        # Apply softmax to get the attention weights
        attention_weights = F.softmax(attn_scores, dim=-1)  # (batch_size, seq_len, seq_len)

        # Compute the output as a weighted sum of the values
        output = torch.matmul(attention_weights, V)  # (batch_size, seq_len, embed_dim)

        return output, attention_weights



batch_size = 1
seq_len = 4
embed_dim = 8

dummy_input = torch.rand(batch_size, seq_len, embed_dim)

attention_layer = SelfAttention(embed_dim)
output, attention_weights = attention_layer(dummy_input)

print("Output shape:", output.shape)  # Should be (batch_size, seq_len, embed_dim)
print("Matrix shape (4 x 4):\n",  attention_weights[0].detach().numpy().round(2))
```
